```python
data = pd.read_excel('%s/ubicaciones.xlsx' % (ruta_lectura))
print(data.columns)
data = data[['n_radicado', 'Latitud', 'Longitud']]
print(data.shape)
dict_op = data.to_dict('list')
print(len(dict_op['n_radicado']))

no_radicado_original = dict_op['n_radicado']
print(len(no_radicado_original))
list_eliminar = [i for i,j in enumerate(dict_op['Latitud']) if pd.isna(j)==True]
print(list_eliminar)
dict_op = {k:[j for i,j in enumerate(v) if i not in list_eliminar] for k,v in dict_op.items()}
print(len(dict_op['Latitud']))
```

```python
def transform_reference(list_op, no):
  list_op = [str(i) for i in list_op]
  list_op = [i.replace('.','') for i in list_op]
  list_op = [i.replace(',','') for i in list_op]
  list_op = [i[:no] + "." + i[no:] for i in list_op]
  print(list_op)
  return list_op
```

```python
list_latitud = transform_reference(list_op=dict_op['Latitud'], no=1)
list_longitud = transform_reference(list_op=dict_op['Longitud'], no=3)
```

```python
list_eliminar = [i for i,j in enumerate(list_latitud) if '-' in j]
list_eliminar = list_eliminar + [i for i,j in enumerate(list_longitud) if '-' not in j]
list_eliminar = list(set(list_eliminar))
print(list_eliminar)
print(len(list_eliminar))
```

```python
def filtrar_valores(list_op):
  list_op = [j for i,j in enumerate(list_op) if i not in list_eliminar]
  return list_op
```

```python
# list_dir = filtrar_valores(list_op=dict_op['Direccion'])
# print(len(list_dir))
list_lat = filtrar_valores(list_op=list_latitud)
print(len(list_lat))
list_lon = filtrar_valores(list_op=list_longitud)
print(len(list_lon))
list_r = filtrar_valores(list_op=dict_op['n_radicado'])
print(len(list_r))
print(list_r)
p = list(set(list_r))
dict_dict = {i:list_r.count(i) for i in p}
dict_dict = {k: v for k, v in sorted(dict_dict.items(), key=lambda item: item[1])}
print(dict_dict)
```

```python
dict_op1 = {r:[list_lat[i], list_lon[i]] for i,r in enumerate(list_r)}
# print(len(dict_op1))
# print(dict_op1)
# c=0
# for i,r in enumerate(list_r):
#   c+=1
#   print(i,r)
# print(c)
dict_op = {}
for r in no_radicado_original:
  if r in dict_op1.keys():
    dict_op[r] = dict_op1[r]
  else:
    dict_op[r] = [0,0]
print(dict_op)
print(len(dict_op))
```

```python
dict_imprimir = {'n_radicado': list(dict_op.keys()),
                 'Latitud': [v[0] for v in list(dict_op.values())],
                 'Longitud': [v[1] for v in list(dict_op.values())]}
df = pd.DataFrame(dict_imprimir)
df
```
